# Uninstalling Anaconda

Follow this notebook **only if you previously installed Anaconda** on your Azure virtual machine. If you have not installed Anaconda, go directly to the [Installation notebook](0_installation.ipynb).

Run all commands in the terminal of your **remote Azure virtual machine**.

## 1. Uninstall Anaconda

> ⚠️ This removes Anaconda and its environments, including `hmd`. Your `HMD-Lab` repository is not removed.

If a Conda environment is active, run `conda deactivate` until no environment name appears in the terminal prompt.

Run the Anaconda uninstall script:

```shell
~/anaconda3/uninstall.sh
```

When prompted to confirm the removal of Anaconda, type `y` and press Enter. Wait until the uninstallation is complete.

## 2. Remove the installer and clear the pip cache

```shell
rm -f ~/Anaconda3-*-Linux-x86_64.sh
rm -rf ~/.cache/pip
```

Do not delete the entire `~/.cache` directory, as it may contain downloaded models.

Close and reopen the terminal, then reconnect to your Azure virtual machine via SSH (or open a new remote terminal in VS Code).

## 3. Install Miniconda

Return to the [Installation notebook](0_installation.ipynb) and follow the instructions from the Miniconda installation section. If you have already cloned the repository, you do not need to clone it again.
