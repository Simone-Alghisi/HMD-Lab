## Synthesis

A dialogue is a list of turns. Each turn is a dictionary with the `voice_id` of the speaker
and the `text` to synthesize:

```python
{"voice_id": VOICE_A, "text": "[rushed] Hi, I need to change my train to Milan."}
```

The function below sends the dialogue to ElevenLabs, saves the audio and plays it.

You can also choose the model and the **stability** of the generation
(`0.0` = Creative, `0.5` = Natural, `1.0` = Robust): uncomment those parameters if an AudioTag
does not seem to have any effect.

````python
from IPython.display import Audio, display


def synthesize(dialogue, filename, stability=0.5):
    audio = client.text_to_dialogue.convert(
        inputs=dialogue,
        model_id="eleven_v3",
        # Lower stability generally gives tags more influence.
        # Remove this parameter if your installed SDK version rejects it.
        settings={
            "stability": stability,
        },
        # output_format="mp3_44100_128",
    )
    save(audio, filename)
    print("Saved:", filename)
    display(Audio(filename))
````

Quick test to check that the key and the voices work:

````python
test = [
    {"voice_id": VOICE_A, "text": "Hello, this is a test of the first voice."},
    {"voice_id": VOICE_B, "text": "And this is the second voice."},
]

synthesize(test, "test.mp3")
````

## AudioTags

AudioTags are natural-language instructions in square brackets that guide the delivery of the
generated speech. You write them inside the text of a turn, and you can combine more than one.

- **Emotions**: `[sad]`, `[angry]`, `[excited]`, `[tired]`, `[frustrated]`, `[relieved]`, `[calm]`
- **Timing**: `[pause]`, `[long pause]`, `[breathes]`, `[rushed]`, `[slow down]`, `[overlapping]`, `[interrupting]`
- **Delivery**: `[whisper]`, `[shout]`, `[softly]`, `[firmly]`, `[reassuringly]`
- **Sound effects**: `[applause]`, `[gulps]`, `[sighs]`

Tags are **instructions**, not **evidence** that the phenomenon actually occurred: always check
what you hear.
