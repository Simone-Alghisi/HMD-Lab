```python
# Instructor solution — connect ASR, the LLM and TTS, then update history.
def spoken_agent(audio, system_prompt, history):
    transcript = transcribe(audio)
    reply = qwen_reply(transcript, system_prompt, history)
    speech, speech_sr = synthesize(reply)
    updated_history = history + [
        {"role": "user", "content": transcript},
        {"role": "assistant", "content": reply},
    ]
    return {"transcript": transcript, "reply": reply,
            "speech": speech, "speech_sr": speech_sr,
            "history": updated_history}
```
